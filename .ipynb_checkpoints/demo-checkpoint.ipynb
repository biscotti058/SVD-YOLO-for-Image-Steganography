## 1. Introduction & Theory

### What is Steganography?
Steganography is the practice of hiding secret information within ordinary, non-secret data (like images) in a way that prevents detection.

### Singular Value Decomposition (SVD)
Given a matrix $A \in \mathbb{R}^{m \times n}$, SVD decomposes it as:

$$A = U \Sigma V^T$$

where:
- $U \in \mathbb{R}^{m \times m}$ — left singular vectors (orthogonal)
- $\Sigma \in \mathbb{R}^{m \times n}$ — diagonal matrix of singular values $\sigma_1 \geq \sigma_2 \geq \ldots \geq 0$
- $V^T \in \mathbb{R}^{n \times n}$ — right singular vectors (orthogonal)

### Why SVD for Steganography?
- **Stability**: Small perturbations to singular values cause small changes in the image
- **Energy compaction**: Most image energy is concentrated in the first few singular values
- **Robustness**: Singular values are resistant to common image attacks (compression, noise)

### Embedding Process
Given cover image $C$ and secret image $S$:

$$C = U_C \Sigma_C V_C^T, \quad S = U_S \Sigma_S V_S^T$$

The stego image is created by:

$$\Sigma_{stego} = \Sigma_C + \alpha \cdot \Sigma_S$$

$$I_{stego} = U_C \cdot \Sigma_{stego} \cdot V_C^T$$

where $\alpha$ is a small scaling factor controlling the trade-off between **imperceptibility** and **robustness**.

### Role of YOLO
YOLO (You Only Look Once) detects objects in the cover image. Complex, textured regions near detected objects are better hiding spots because the human visual system is less sensitive to changes in these areas.

## 3. Generate Sample Images

We'll create sample cover and secret images for demonstration.
You can also use your own images by placing them in the `images/` folder.

```python
def generate_sample_cover(size=(512, 512)):
    """Generate a sample cover image with diverse textures."""
    h, w = size
    img = np.zeros((h, w, 3), dtype=np.uint8)
    
    # Sky gradient (top)
    for y in range(h // 3):
        ratio = y / (h // 3)
        img[y, :] = [int(135 + 50 * ratio), int(206 + 30 * ratio), int(235 + 20 * ratio)]
    
    # Green hills (middle)
    for y in range(h // 3, 2 * h // 3):
        ratio = (y - h // 3) / (h // 3)
        base_g = int(160 - 40 * ratio)
        for x in range(w):
            noise = np.random.randint(-15, 15)
            img[y, x] = [34 + noise, base_g + noise, 34 + noise]
    
    # Complex pattern (bottom) - bricks
    for y in range(2 * h // 3, h):
        for x in range(w):
            by = (y - 2 * h // 3) % 20
            bx = x % 40
            if by < 2 or (bx < 2 and by > 10):
                img[y, x] = [100, 100, 100]  # mortar
            else:
                noise = np.random.randint(-10, 10)
                img[y, x] = [180 + noise, 100 + noise, 60 + noise]  # brick
    
    # Add some circles (objects)
    cv2.circle(img, (w // 4, h // 4), 40, (255, 220, 50), -1)  # sun
    cv2.rectangle(img, (w // 2, h // 3), (w // 2 + 80, 2 * h // 3), (100, 80, 60), -1)  # tree trunk
    cv2.circle(img, (w // 2 + 40, h // 4 + 20), 60, (34, 139, 34), -1)  # tree crown
    
    return img


def generate_sample_secret(size=(256, 256)):
    """Generate a sample secret image (a simple pattern with text)."""
    img = np.ones((size[0], size[1], 3), dtype=np.uint8) * 255
    
    # Draw a secret pattern
    cv2.putText(img, 'SECRET', (20, size[0] // 2), 
                cv2.FONT_HERSHEY_SIMPLEX, 2, (255, 0, 0), 4)
    cv2.putText(img, 'MESSAGE', (10, size[0] // 2 + 60),
                cv2.FONT_HERSHEY_SIMPLEX, 1.5, (0, 0, 255), 3)
    
    # Add geometric shapes
    cv2.rectangle(img, (10, 10), (size[1]-10, size[0]-10), (0, 0, 0), 3)
    cv2.circle(img, (size[1]//2, size[0]//4), 30, (0, 128, 0), -1)
    
    return img


# Generate or load images
cover_path = 'images/cover.png'
secret_path = 'images/secret.png'

if not os.path.exists(cover_path):
    cover = generate_sample_cover()
    save_image(cover, cover_path)
    print(f"Generated sample cover image: {cover_path}")
else:
    cover = load_image(cover_path)
    print(f"Loaded cover image: {cover_path}")

if not os.path.exists(secret_path):
    secret = generate_sample_secret()
    save_image(secret, secret_path)
    print(f"Generated sample secret image: {secret_path}")
else:
    secret = load_image(secret_path)
    print(f"Loaded secret image: {secret_path}")

# Display
fig, axes = plt.subplots(1, 2, figsize=(12, 5))
axes[0].imshow(cover)
axes[0].set_title(f'Cover Image\n{cover.shape}')
axes[0].axis('off')
axes[1].imshow(secret)
axes[1].set_title(f'Secret Image\n{secret.shape}')
axes[1].axis('off')
plt.tight_layout()
plt.savefig('output/01_input_images.png', bbox_inches='tight')
plt.show()
```

## 4. Basic SVD Steganography

Let's start with basic SVD steganography (no YOLO) to understand the core algorithm.

```python
# Basic SVD Steganography
alpha = 0.01
stego_basic = SVDSteganography(alpha=alpha)

# Encode
stego_img = stego_basic.encode(cover, secret)

# Compute metrics
psnr_val = compute_psnr(cover, stego_img)
ssim_val = compute_ssim(cover, stego_img)

print(f"Alpha: {alpha}")
print(f"PSNR: {psnr_val:.2f} dB")
print(f"SSIM: {ssim_val:.4f}")

# Decode
recovered = stego_basic.decode(stego_img)

# Visualize
fig, axes = plt.subplots(2, 2, figsize=(14, 12))

axes[0, 0].imshow(cover)
axes[0, 0].set_title('Cover Image (Original)')
axes[0, 0].axis('off')

axes[0, 1].imshow(stego_img)
axes[0, 1].set_title(f'Stego Image\nPSNR={psnr_val:.1f}dB, SSIM={ssim_val:.4f}')
axes[0, 1].axis('off')

# Difference (amplified for visibility)
diff = np.abs(cover.astype(float) - stego_img.astype(float))
diff_amplified = np.clip(diff * 20, 0, 255).astype(np.uint8)
axes[1, 0].imshow(diff_amplified)
axes[1, 0].set_title('Difference (20x amplified)')
axes[1, 0].axis('off')

axes[1, 1].imshow(recovered)
axes[1, 1].set_title('Recovered Secret')
axes[1, 1].axis('off')

plt.suptitle(f'Basic SVD Steganography (α = {alpha})', fontsize=16, fontweight='bold')
plt.tight_layout()
plt.savefig('output/02_basic_svd.png', bbox_inches='tight')
plt.show()
```

## 5. Understanding SVD Decomposition

Let's visualize how SVD decomposes an image and how different numbers of singular values reconstruct it.

```python
# SVD decomposition visualization
gray_cover = cv2.cvtColor(cover, cv2.COLOR_RGB2GRAY).astype(np.float64)
U, S, Vt = np.linalg.svd(gray_cover, full_matrices=False)

# Plot singular value spectrum
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# Singular values (log scale)
axes[0].semilogy(S, 'b-', linewidth=2)
axes[0].set_xlabel('Index')
axes[0].set_ylabel('Singular Value (log scale)')
axes[0].set_title('Singular Value Spectrum')
axes[0].grid(True, alpha=0.3)

# Cumulative energy
cumulative_energy = np.cumsum(S**2) / np.sum(S**2)
axes[1].plot(cumulative_energy, 'r-', linewidth=2)
axes[1].axhline(y=0.95, color='g', linestyle='--', label='95% energy')
axes[1].axhline(y=0.99, color='orange', linestyle='--', label='99% energy')
k_95 = np.argmax(cumulative_energy >= 0.95)
k_99 = np.argmax(cumulative_energy >= 0.99)
axes[1].axvline(x=k_95, color='g', linestyle=':', alpha=0.5)
axes[1].axvline(x=k_99, color='orange', linestyle=':', alpha=0.5)
axes[1].set_xlabel('Number of Singular Values (k)')
axes[1].set_ylabel('Cumulative Energy Ratio')
axes[1].set_title(f'Energy: 95% at k={k_95}, 99% at k={k_99}')
axes[1].legend()
axes[1].grid(True, alpha=0.3)

# Reconstruction with different ranks
ranks = [1, 5, 10, 25, 50, 100]
axes[2].set_visible(False)

plt.tight_layout()
plt.savefig('output/03_svd_spectrum.png', bbox_inches='tight')
plt.show()

# Low-rank reconstructions
fig, axes = plt.subplots(2, 3, figsize=(16, 10))
for idx, k in enumerate(ranks):
    row, col = idx // 3, idx % 3
    reconstructed = U[:, :k] @ np.diag(S[:k]) @ Vt[:k, :]
    reconstructed = np.clip(reconstructed, 0, 255).astype(np.uint8)
    axes[row, col].imshow(reconstructed, cmap='gray')
    energy = np.sum(S[:k]**2) / np.sum(S**2) * 100
    axes[row, col].set_title(f'Rank {k}\nEnergy: {energy:.1f}%')
    axes[row, col].axis('off')

plt.suptitle('Low-Rank SVD Approximations of Cover Image', fontsize=16, fontweight='bold')
plt.tight_layout()
plt.savefig('output/04_low_rank_approx.png', bbox_inches='tight')
plt.show()
```

## 6. YOLO-Guided Region Selection

YOLO detects objects in the cover image. Regions with detected objects typically have complex textures where hidden data is less noticeable.

> **Note**: This section requires the `ultralytics` package and will download the YOLOv8 model on first run.
> If YOLO is not available, a texture-based fallback is used.

```python
# YOLO Region Selection
try:
    yolo_selector = YOLORegionSelector(model_name='yolov8n.pt', confidence=0.25)
    detections = yolo_selector.detect_objects(cover)
    
    print(f"YOLO Detections: {len(detections)}")
    for det in detections:
        print(f"  - {det['class']}: confidence={det['confidence']:.2f}, bbox={det['bbox']}")
    
    # Generate masks
    yolo_mask = yolo_selector.generate_embedding_mask(cover)
    yolo_only_mask = yolo_selector.generate_embedding_mask(cover, use_texture=False)
    texture_mask = generate_texture_only_mask(cover)
    
    # Visualize
    fig, axes = plt.subplots(2, 2, figsize=(14, 12))
    
    axes[0, 0].imshow(cover)
    axes[0, 0].set_title('Original Cover Image')
    axes[0, 0].axis('off')
    
    axes[0, 1].imshow(texture_mask, cmap='hot')
    axes[0, 1].set_title('Texture Complexity Map')
    axes[0, 1].axis('off')
    
    # YOLO mask visualization
    mask_vis = yolo_selector.visualize_mask(cover, yolo_mask, detections)
    axes[1, 0].imshow(mask_vis)
    axes[1, 0].set_title(f'YOLO Detections + Mask\n({len(detections)} objects detected)')
    axes[1, 0].axis('off')
    
    axes[1, 1].imshow(yolo_mask, cmap='hot')
    axes[1, 1].set_title('Combined Embedding Mask\n(YOLO + Texture)')
    axes[1, 1].axis('off')
    
    plt.suptitle('YOLO-Guided Region Selection', fontsize=16, fontweight='bold')
    plt.tight_layout()
    plt.savefig('output/05_yolo_regions.png', bbox_inches='tight')
    plt.show()
    
    YOLO_AVAILABLE = True

except Exception as e:
    print(f"YOLO not available: {e}")
    print("Using texture-based fallback...")
    
    texture_mask = generate_texture_only_mask(cover)
    
    fig, axes = plt.subplots(1, 2, figsize=(12, 5))
    axes[0].imshow(cover)
    axes[0].set_title('Cover Image')
    axes[0].axis('off')
    axes[1].imshow(texture_mask, cmap='hot')
    axes[1].set_title('Texture Complexity Map (Fallback)')
    axes[1].axis('off')
    plt.tight_layout()
    plt.savefig('output/05_texture_regions.png', bbox_inches='tight')
    plt.show()
    
    YOLO_AVAILABLE = False
```

## 8. Comparative Analysis

Let's compare different embedding strengths (alpha) and observe the trade-off between **imperceptibility** (PSNR/SSIM) and **recovery quality**.

```python
# Alpha comparison
alphas = [0.001, 0.005, 0.01, 0.02, 0.05, 0.1]
psnr_values = []
ssim_values = []
recovery_psnrs = []

fig_stego, axes_stego = plt.subplots(2, 3, figsize=(18, 12))
fig_recov, axes_recov = plt.subplots(2, 3, figsize=(18, 12))

for idx, alpha in enumerate(alphas):
    row, col = idx // 3, idx % 3
    
    stego_engine = SVDSteganography(alpha=alpha)
    stego_result = stego_engine.encode(cover, secret)
    recovered_result = stego_engine.decode(stego_result)
    
    psnr_val = compute_psnr(cover, stego_result)
    ssim_val = compute_ssim(cover, stego_result)
    
    # Recovery quality
    secret_resized = stego_engine._resize_secret(cover, secret)
    rec_psnr = compute_psnr(secret_resized, recovered_result)
    
    psnr_values.append(psnr_val)
    ssim_values.append(ssim_val)
    recovery_psnrs.append(rec_psnr)
    
    # Stego images
    axes_stego[row, col].imshow(stego_result)
    axes_stego[row, col].set_title(f'α={alpha}\nPSNR={psnr_val:.1f}dB, SSIM={ssim_val:.4f}')
    axes_stego[row, col].axis('off')
    
    # Recovered images
    axes_recov[row, col].imshow(recovered_result)
    axes_recov[row, col].set_title(f'α={alpha}\nRecovery PSNR={rec_psnr:.1f}dB')
    axes_recov[row, col].axis('off')

fig_stego.suptitle('Stego Images at Different Alpha Values', fontsize=16, fontweight='bold')
fig_stego.tight_layout()
fig_stego.savefig('output/07_alpha_comparison_stego.png', bbox_inches='tight')

fig_recov.suptitle('Recovered Secrets at Different Alpha Values', fontsize=16, fontweight='bold')
fig_recov.tight_layout()
fig_recov.savefig('output/07_alpha_comparison_recovered.png', bbox_inches='tight')

plt.show()
```

```python
# Plot PSNR/SSIM vs Alpha trade-off
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# PSNR of stego image
axes[0].plot(alphas, psnr_values, 'bo-', linewidth=2, markersize=8)
axes[0].axhline(y=30, color='r', linestyle='--', alpha=0.7, label='30dB threshold')
axes[0].set_xlabel('Alpha (α)')
axes[0].set_ylabel('PSNR (dB)')
axes[0].set_title('Cover vs Stego: PSNR')
axes[0].set_xscale('log')
axes[0].legend()
axes[0].grid(True, alpha=0.3)

# SSIM of stego image
axes[1].plot(alphas, ssim_values, 'go-', linewidth=2, markersize=8)
axes[1].axhline(y=0.95, color='r', linestyle='--', alpha=0.7, label='0.95 threshold')
axes[1].set_xlabel('Alpha (α)')
axes[1].set_ylabel('SSIM')
axes[1].set_title('Cover vs Stego: SSIM')
axes[1].set_xscale('log')
axes[1].legend()
axes[1].grid(True, alpha=0.3)

# Recovery PSNR
axes[2].plot(alphas, recovery_psnrs, 'ro-', linewidth=2, markersize=8)
axes[2].set_xlabel('Alpha (α)')
axes[2].set_ylabel('Recovery PSNR (dB)')
axes[2].set_title('Secret Recovery Quality')
axes[2].set_xscale('log')
axes[2].grid(True, alpha=0.3)

plt.suptitle('Trade-off Analysis: Imperceptibility vs Recovery Quality', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.savefig('output/08_tradeoff_analysis.png', bbox_inches='tight')
plt.show()

# Summary table
print(f"{'Alpha':<10} {'PSNR(dB)':<12} {'SSIM':<10} {'Recovery PSNR':<15}")
print('-' * 47)
for a, p, s, rp in zip(alphas, psnr_values, ssim_values, recovery_psnrs):
    print(f"{a:<10.3f} {p:<12.2f} {s:<10.4f} {rp:<15.2f}")
```

## 9. Robustness Analysis

Let's test how robust the steganography is against common image attacks:
- **JPEG Compression**
- **Gaussian Noise**
- **Gaussian Blur**
- **Scaling (resize)**

```python
import io

def jpeg_compress(image: np.ndarray, quality: int = 50) -> np.ndarray:
    """Apply JPEG compression to an image."""
    pil_img = Image.fromarray(image)
    buffer = io.BytesIO()
    pil_img.save(buffer, format='JPEG', quality=quality)
    buffer.seek(0)
    return np.array(Image.open(buffer).convert('RGB'))


def add_gaussian_noise(image: np.ndarray, std: float = 10) -> np.ndarray:
    """Add Gaussian noise to an image."""
    noise = np.random.normal(0, std, image.shape)
    noisy = np.clip(image.astype(float) + noise, 0, 255).astype(np.uint8)
    return noisy


def apply_blur(image: np.ndarray, ksize: int = 5) -> np.ndarray:
    """Apply Gaussian blur."""
    return cv2.GaussianBlur(image, (ksize, ksize), 0)


def scale_attack(image: np.ndarray, scale: float = 0.5) -> np.ndarray:
    """Scale down and up (lossy resize)."""
    h, w = image.shape[:2]
    small = cv2.resize(image, (int(w * scale), int(h * scale)))
    return cv2.resize(small, (w, h))


# Create stego image for robustness testing
alpha = 0.01
stego_engine = SVDSteganography(alpha=alpha)
stego_img = stego_engine.encode(cover, secret)

# Apply attacks
attacks = {
    'Original Stego': stego_img,
    'JPEG Q=50': jpeg_compress(stego_img, quality=50),
    'JPEG Q=20': jpeg_compress(stego_img, quality=20),
    'Noise σ=5': add_gaussian_noise(stego_img, std=5),
    'Noise σ=15': add_gaussian_noise(stego_img, std=15),
    'Blur k=5': apply_blur(stego_img, ksize=5),
    'Scale 0.5x': scale_attack(stego_img, scale=0.5),
    'Scale 0.25x': scale_attack(stego_img, scale=0.25),
}

# Attempt recovery after each attack
fig, axes = plt.subplots(2, 4, figsize=(20, 10))
attack_results = []

for idx, (name, attacked_img) in enumerate(attacks.items()):
    row, col = idx // 4, idx % 4
    
    # Try to decode
    try:
        recovered_attacked = stego_engine.decode(attacked_img)
        secret_resized = stego_engine._resize_secret(cover, secret)
        rec_psnr = compute_psnr(secret_resized, recovered_attacked)
        rec_ssim = compute_ssim(secret_resized, recovered_attacked)
    except Exception as e:
        rec_psnr = 0
        rec_ssim = 0
    
    # Stego quality
    stego_psnr = compute_psnr(stego_img, attacked_img)
    
    attack_results.append({
        'attack': name,
        'stego_psnr': stego_psnr,
        'recovery_psnr': rec_psnr,
        'recovery_ssim': rec_ssim
    })
    
    axes[row, col].imshow(recovered_attacked if rec_psnr > 0 else attacked_img)
    axes[row, col].set_title(f'{name}\nRec. PSNR={rec_psnr:.1f}dB', fontsize=10)
    axes[row, col].axis('off')

plt.suptitle('Robustness: Recovered Secrets After Attacks', fontsize=16, fontweight='bold')
plt.tight_layout()
plt.savefig('output/09_robustness.png', bbox_inches='tight')
plt.show()

# Summary table
print(f"{'Attack':<20} {'Stego PSNR':<14} {'Recovery PSNR':<16} {'Recovery SSIM':<14}")
print('-' * 64)
for r in attack_results:
    print(f"{r['attack']:<20} {r['stego_psnr']:<14.2f} {r['recovery_psnr']:<16.2f} {r['recovery_ssim']:<14.4f}")
```

## 10. Conclusions

### Key Findings

1. **SVD Effectiveness**: SVD-based steganography successfully hides secret images within cover images with minimal visual distortion (PSNR > 30dB at α ≤ 0.01).

2. **Alpha Trade-off**: There is a clear trade-off between imperceptibility and recovery quality:
   - Low α (≤ 0.005): Excellent imperceptibility but fragile recovery
   - Medium α (0.01-0.02): Good balance
   - High α (≥ 0.05): Robust recovery but visible artifacts

3. **YOLO Enhancement**: YOLO-guided region selection directs embedding towards complex, textured areas where modifications are less perceptible to the human eye.

4. **Robustness**: The method shows moderate robustness to:
   - JPEG compression (especially at higher quality)
   - Mild noise addition
   - Less robust against aggressive scaling and heavy compression

### Limitations

- **Decoding keys**: Extracting the secret requires the `keys.npz` file (U, V matrices from SVD). The recipient must receive the keys; this is not "blind" steganography (password/seed only).
- **Robustness**: Aggressive JPEG compression, heavy resizing, and strong noise can degrade or destroy the message.
- **Hiding only**: Data is not encrypted; anyone who extracts the payload from the SVD sees the secret in the clear.

### Future developments

- **Blind steganography**: recover the message with only a password/seed, without transmitting SVD matrices.
- **Encryption** (e.g. AES) of the payload before embedding, for confidentiality even if the payload is extracted.
- **YOLO model choice**: use heavier models (yolov8l, yolov8x) with GPU for more accurate masks.
- **Adaptive alpha**: local adaptation of α based on region complexity.
- Text steganography (hiding text messages instead of images).
- Testing on larger real-world datasets.
